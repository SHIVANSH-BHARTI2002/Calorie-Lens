```python
import os
import threading
from icrawler.builtin import BingImageCrawler


# ══════════════════════════════════════════════════════════
#  CONFIG  —  edit this section only
# ══════════════════════════════════════════════════════════

OUTPUT_DIR  = "testing_dataset"     # root folder for all downloads
IMAGES_EACH = 250           # images to download per query

# Format:
#   "sub-folder": ["query 1", "query 2", ...]

QUERIES = {
"mysore_pak": ["mysore pak on plate", "mysore pak sweet on plate"],
"pootharekulu": ["pootharekulu sweet on plate", "pootharekulu dessert on plate"],
"sohan_halwa": ["sohan halwa on plate", "sohan halwa sweet on plate"],
"sutar_feni": ["sutar feni on plate", "sutar feni sweet on plate"]
}

# ══════════════════════════════════════════════════════════


def _crawl(query, folder, max_num):
    os.makedirs(folder, exist_ok=True)
    existing = len(os.listdir(folder))

    BingImageCrawler(
        feeder_threads=2,
        parser_threads=2,
        downloader_threads=4,
        storage={"root_dir": folder}
    ).crawl(
        keyword=query,
        max_num=max_num,
        min_size=(100, 100),
        file_idx_offset=existing      # avoids overwriting previous downloads
    )


def _count(folder):
    exts = {".jpg", ".jpeg", ".png", ".webp"}
    return sum(1 for f in os.listdir(folder) if os.path.splitext(f)[1].lower() in exts)


def run():
    total_queries = sum(len(q) for q in QUERIES.values())

    print(f"\n{'─'*52}")
    print(f"  Output   : {OUTPUT_DIR}/")
    print(f"  Classes  : {len(QUERIES)}")
    print(f"  Queries  : {total_queries}")
    print(f"  Per query: {IMAGES_EACH} images")
    print(f"  Expected : ~{total_queries * IMAGES_EACH} total")
    print(f"{'─'*52}\n")

    threads = [
        threading.Thread(
            target=_crawl,
            args=(query, os.path.join(OUTPUT_DIR, subfolder), IMAGES_EACH),
            name=query
        )
        for subfolder, queries in QUERIES.items()
        for query in queries
    ]

    for t in threads:
        print(f"[START]  {t.name}")
        t.start()

    for t in threads:
        t.join()

    # ── Summary ──────────────────────────────────────────
    print(f"\n{'─'*52}")
    print("  SUMMARY")
    print(f"{'─'*52}")

    total = 0
    for subfolder in sorted(QUERIES):
        path = os.path.join(OUTPUT_DIR, subfolder)
        n = _count(path) if os.path.exists(path) else 0
        total += n
        print(f"  {subfolder:<30} {n:>4} images")

    print(f"{'─'*52}")
    print(f"  {'TOTAL':<30} {total:>4} images")
    print(f"{'─'*52}\n")


if __name__ == "__main__":
    run()
```
